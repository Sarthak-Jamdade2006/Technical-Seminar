```python
def get_combined_mask(folder, image_name):
    base_name = os.path.splitext(image_name)[0]

    mask_files = [
        f for f in os.listdir(folder)
        if f.startswith(base_name + "_mask")
        and f.endswith(".png")
    ]

    if not mask_files:
        return None

    combined_mask = None

    for mask_file in mask_files:
        mask = np.array(Image.open(os.path.join(folder, mask_file)))

        if combined_mask is None:
            combined_mask = mask
        else:
            combined_mask = np.maximum(combined_mask, mask)

    return combined_mask
```

```python
LABELS = {
    "benign": 0,
    "malignant": 1
}

print(LABELS)
```

```python
samples = []

for class_name, label in LABELS.items():
    folder = os.path.join("BUSI", class_name)

    for file_name in os.listdir(folder):
        if file_name.endswith(".png") and "_mask" not in file_name:
            samples.append({
                "image_path": os.path.join(folder, file_name),
                "class_name": class_name,
                "label": label
            })

print("Total samples:", len(samples))
print("First 5 samples:")

for sample in samples[:5]:
    print(sample)
```

```python
from collections import Counter

label_counts = Counter(sample["label"] for sample in samples)

print("Benign:", label_counts[0])
print("Malignant:", label_counts[1])
```

```python
sample_image = Image.open(samples[0]["image_path"]).convert("RGB")

print("Image size:", sample_image.size)
print("Image mode:", sample_image.mode)
```

```python
resized_image = sample_image.resize((224, 224))

print("Resized image size:", resized_image.size)
```

```python
image_array = np.array(resized_image)

print("Array shape:", image_array.shape)
print("Data type:", image_array.dtype)
```

```python
image_array = image_array.astype(np.float32) / 255.0

print("Min:", image_array.min())
print("Max:", image_array.max())
```

```python
import torch

image_tensor = torch.tensor(image_array).permute(2, 0, 1)

print("Tensor shape:", image_tensor.shape)
```

```python
import torchvision.models as models

teacher = models.vgg16(weights="DEFAULT")
print("VGG16 loaded successfully.")
```

```python
teacher.eval()

print("Teacher ready:", not teacher.training)
```

```python
import torch.nn as nn

teacher.classifier[6] = nn.Linear(4096, 2)

print("Output layer:", teacher.classifier[6])
```

```python
print("Current tensor shape:", image_tensor.shape)
```

```python
image_tensor = image_tensor.unsqueeze(0)
print("Fixed tensor shape:", image_tensor.shape)
```

```python
teacher.eval()

with torch.no_grad():
    output = teacher(image_tensor)

print("Teacher output shape:", output.shape)
```

```python
student = models.vgg16(weights="DEFAULT")
student.classifier[6] = nn.Linear(4096, 2)

print("Student ready:", student.classifier[6])
```

```python
student.eval()

with torch.no_grad():
    student_output = student(image_tensor)

print("Student output shape:", student_output.shape)
```

```python
criterion = nn.CrossEntropyLoss()

label_tensor = torch.tensor([samples[0]["label"]], dtype=torch.long)

teacher_loss = criterion(output, label_tensor)
student_loss = criterion(student_output, label_tensor)

print("Teacher loss:", teacher_loss.item())
print("Student loss:", student_loss.item())
```

```python
teacher_pred = torch.argmax(output, dim=1).item()
student_pred = torch.argmax(student_output, dim=1).item()

print("True label:", samples[0]["label"])
print("Teacher prediction:", teacher_pred)
print("Student prediction:", student_pred)
```

```python
with torch.no_grad():
    teacher_features = teacher.avgpool(teacher.features(image_tensor))
    teacher_features = torch.flatten(teacher_features, 1)
    
    student_features = student.avgpool(student.features(image_tensor))
    student_features = torch.flatten(student_features, 1)

print("Teacher features:", teacher_features.shape)
print("Student features:", student_features.shape)
```

```python
print("Teacher feature dimension:", teacher_features.shape[1])
print("Student feature dimension:", student_features.shape[1])
```

```python
teacher_dist = torch.softmax(teacher_features, dim=1)
student_dist = torch.softmax(student_features, dim=1)

print("Teacher distribution sum:", teacher_dist.sum().item())
print("Student distribution sum:", student_dist.sum().item())
```

```python
ddl_loss = torch.nn.functional.kl_div(
    torch.log(student_dist + 1e-8),
    teacher_dist,
    reduction="batchmean"
)

print("DDL loss:", ddl_loss.item())
```

```python
ddl_loss = torch.tensor(0.0)

print("Temporary DDL loss reset:", ddl_loss.item())
```

```python
from sklearn.model_selection import train_test_split

train_samples, test_samples = train_test_split(
    samples,
    test_size=0.20,
    stratify=[s["label"] for s in samples],
    random_state=42
)

print("Train:", len(train_samples))
print("Test:", len(test_samples))
```

```python
class BUSIDataset(torch.utils.data.Dataset):
    def __init__(self, samples):
        self.samples = samples

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        sample = self.samples[idx]

        image = Image.open(sample["image_path"]).convert("RGB")
        image = image.resize((224, 224))

        image = np.array(image).astype(np.float32) / 255.0
        image = torch.tensor(image).permute(2, 0, 1)

        label = torch.tensor(sample["label"], dtype=torch.long)

        return image, label
```

```python
train_dataset = BUSIDataset(train_samples)
test_dataset = BUSIDataset(test_samples)

print("Train dataset:", len(train_dataset))
print("Test dataset:", len(test_dataset))
```

```python
train_loader = torch.utils.data.DataLoader(
    train_dataset,
    batch_size=32,
    shuffle=True
)

test_loader = torch.utils.data.DataLoader(
    test_dataset,
    batch_size=32,
    shuffle=False
)

print("Train batches:", len(train_loader))
print("Test batches:", len(test_loader))
```

```python
images, labels = next(iter(train_loader))

print("Images:", images.shape)
print("Labels:", labels.shape)
```

```python
teacher_optimizer = torch.optim.SGD(
    teacher.parameters(),
    lr=0.001,
    momentum=0.9,
    weight_decay=0.0005
)

print("Teacher optimizer ready")
```

```python
teacher.train()

for images, labels in train_loader:
    teacher_optimizer.zero_grad()

    outputs = teacher(images)
    loss = criterion(outputs, labels)

    loss.backward()
    teacher_optimizer.step()

print("Teacher 1-epoch training completed")
```

```python
teacher.eval()

correct = 0
total = 0

with torch.no_grad():
    for images, labels in test_loader:
        outputs = teacher(images)
        predictions = torch.argmax(outputs, dim=1)

        correct += (predictions == labels).sum().item()
        total += labels.size(0)

teacher_accuracy = correct / total

print("Teacher Test Accuracy:", teacher_accuracy)
```
